### TF-IDF

```python
import nltk
from nltk.tokenize import word_tokenize
from math import log
```

```python
nltk.download('punkt')
```

```python
documents = [
    "I love cats",
    "I love dogs",
    "Cats are amazing pets"
]
```

```python
# Step 1 : Tokenize documents 
tokenized_docs = [word_tokenize(doc.lower()) for doc in documents]
```

```python
# Step 2 : Build vocabulary (all unique words)
vocab = sorted(set(word for doc in tokenized_docs for word in doc))
```

```python
# Step 3: Compute Term Frequency (TF)
def compute_tf(doc_tokens):
    tf_dict = {}
    total_words = len(doc_tokens)
    for word in vocab:
        tf_dict[word] = doc_tokens.count(word) / total_words
    return tf_dict
```

```python
tf_list = [compute_tf(doc) for doc in tokenized_docs]
```

```python
# Step 4: Compute Inverse Document Frequency (IDF)
def compute_idf(tokenized_docs):
    import math
    N = len(tokenized_docs)
    idf_dict = {}
    for word in vocab:
        count = sum(1 for doc in tokenized_docs if word in doc)
        idf_dict[word] = log((N + 1) / (count + 1)) + 1  # Smoothing
    return idf_dict
```

```python
idf_dict = compute_idf(tokenized_docs)
```

```python
# Step 5: Compute TF-IDF
tfidf_list = []
for tf in tf_list:
    tfidf = {word: tf[word] * idf_dict[word] for word in vocab}
    tfidf_list.append(tfidf)
```

```python
# Step 6: Display TF-IDF
for i, tfidf in enumerate(tfidf_list):
    print(f"\nDocument {i+1} TF-IDF:")
    for word, score in tfidf.items():
        if score > 0:
            print(f"{word}: {score:.3f}")
```
